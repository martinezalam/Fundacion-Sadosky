Analisis de Valores nulos

```python
#Primero hago una analisis de los datos para ver que columnas me interesan y cuales no
#Me doy cuenta que hay columnas con valores NaN, por lo que quiero ver que son esas columnas
df.info()
```

```python
#Veo la cantidad de valores nulos por columna
df.isnull().sum()
```

```python
#Elimino las columnas que no me interesan para el analisis
df = df.drop(['Unnamed: 16', 'Unnamed: 17', 'Unnamed: 18', 'Unnamed: 19', 'Unnamed: 20', 'Unnamed: 21', 'Unnamed: 22', 'Unnamed: 23', 'Unnamed: 24', 'Unnamed: 25'], axis=1)
```

```python
#Elimino columnas que no me interesan para el analisis ya que tiene valores nulos
df = df.dropna(how='all')
```

```python
#Chequeo que se hayan eliminado las columnas y las filas nulas
df.info()
```


Analisis de Inconsistencias

```python
#Me fijo si los valores de tvd landing no son iguales a cero
inconsistencias_tvd_landing = df[df['tvd_landing'] == 0]
print("Cantidad de inconsistencias en tvd_landing:", len(inconsistencias_tvd_landing))

#Realizo lo mismo con la columna vol_swater_frac y vol_xlink_frac
Pozos_sin_slickwater = df[df['vol_swater_frac'] == 0]
print("Pozos sin Slickwater:", len(Pozos_sin_slickwater))

Pozos_sin_crosslinker = df[df['vol_xlink_frac'] == 0]
print("Pozos sin Crosslinker:", len(Pozos_sin_crosslinker))

```

```python
#Ahora quiero ver si los pozos tienen valor nulo en vol_xlink_frac tienen algun valor en vol_swater_frac y viceversa
Cruce_Slickwater_Crosslinker = df[(df['vol_xlink_frac'].isnull()) & (df['vol_swater_frac'].notnull()) | (df['vol_xlink_frac'].notnull()) & (df['vol_swater_frac'].isnull())]
print("Inconsistencias:", len(Cruce_Slickwater_Crosslinker))

#Aca me doy cuenta que los pozos que tienen valor nulo en crosslinker tienen valor en slickwater y viceversa
#Por lo que esta bien considerando a nivel de ingenieria
```

```python
#Busco valores de porosidad imposible
errores_porosidad = df[(df['omphi_'] < 0) | (df['omphi_'] > 1)]
print("Cantidad de inconsistencias en porosidad:", len(errores_porosidad))
#Busco valores de profundidad nulos
errores_profundidad = df[df['tvd_landing'].isnull()]
print("Cantidad de inconsistencias en profundidad:", len(errores_profundidad))
#Busco errores de tiempo imposible
errores_tiempo = df[df['frac_y_pem_dias'] < 0]
print("Errores Temporales (Fractura-PEM):", len(errores_tiempo))
#Busco errores de volumenes imposibles
errores_volumen = df[df['210d_kbbl'] < 0]
print("Volúmenes negativos:", len(errores_volumen))

#No veo inconsistencias en los datos, por lo que puedo seguir con el analisis de los mismos
```
