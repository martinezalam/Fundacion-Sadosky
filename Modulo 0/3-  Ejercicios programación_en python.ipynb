### 12 - Producción diaria de un pozo

Una empresa registra la producción diaria de petróleo (en barriles) durante una semana.

````python
produccion = [520, 540, 510, 600, 590, 580, 610]
produccion
````

Consignas

* Calcular la producción total.
* Calcular la producción promedio.
* Encontrar la producción máxima y mínima.
* Determinar cuántos días la producción superó los 550 barriles.

````python
# Ejecutar aquí las instrucciones de código y encontrá la respuesta
produccion = [520, 540, 510, 600, 590, 580, 610]

total = sum(produccion)

promedio = total / len(produccion)

maximo = max(produccion)

minimo = min(produccion)

dias_superiores = 0

for valor in produccion:
    if valor > 550:
        dias_superiores += 1

print("Producción total:", total)
print("Promedio:", promedio)
print("Máxima:", maximo)
print("Mínima:", minimo)
print("Días sobre 550 barriles:", dias_superiores)
````
