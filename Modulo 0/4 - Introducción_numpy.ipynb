# 🎯 Introducción a Numpy



## Importar Numpy
La convención más usada es importar NumPy con el alias np.

```python
import numpy as np
```

## Crear un ndarray

Desde una lista

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print(arr)
```

Arreglo bidimensional

```python
matriz = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(matriz)
```

## Propiedades importantes

Dimensiones

```python
print(matriz.ndim)
```

Forma (shape)

```python
print(matriz.shape)
```

Significa:

2 filas
3 columnas

Cantidad de elementos (size)

```python
print(matriz.size)
```

Tipo de dato (dtype)

```python
print(matriz.dtype)
```

## Crear arreglos rápidamente

Arreglo de ceros

```python
np.zeros((3, 3))
```

Arreglo de unos

```python
np.ones((2, 4))
```

Secuencia numérica

```python
np.arange(0, 10, 2)
```

Valores espaciados uniformemente

```python
np.linspace(0, 1, 5)
```

## Operaciones básicas

Suma

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b) # Suma
```

Multiplicación

```python
print(a * b)
```

Operaciones con escalares

```python
print(a * 10)
```

## Indexación y slicing

Acceder a elementos

```python
print(a[0]) #nos busca en la variable a el valor que esta en la posicion 0
```

Subconjuntos

```python
print(a[1:]) #nos deja ver los valores que estan desde la posicion 1 en adelante
```

En matrices

```python
print(matriz[0, 1])
```

## Más de 2 dimensiones

```python
arr_5D = np.zeros((2, 3, 4, 5, 6))

print(arr_5D.shape)
```

Este arreglo tiene:

* 2 bloques
* 3 sub-bloques
* 4 matrices
* 5 filas
* 6 columnas

```python
arr_5D
```

Acceder a un elemento

```python
print(arr_5D[0, 1, 2, 3, 4])
```

```python
arr_5D[0,0,0,:,:] # matriz de 5 filas y 6 columnas en posicion 0,0,0
```

Ejemplo numpy

```python
import numpy as np

# Crear arreglo
datos = np.array([10, 20, 30, 40])

# Operaciones
print("Suma:", datos.sum())
print("Promedio:", datos.mean())
print("Máximo:", datos.max())

# Multiplicación
print("Datos x2:", datos * 2)
```

📝 **Ejercicio**

Ejecutar el siguiente código:

```python
import numpy as np

arr_empresas = np.array([
    [2008, 16, "ALIANZA PETROLERA ARGENTINA S.A.", "2008-11-28"],
    [2008, 28, "APCO ARGENTINA INC.", "2008-11-28"],
    [2008, 1194, "BLACK GOLD S.A.", "2009-01-19"]
])

arr_empresas
```

```python
arr_empresas[:, 2]      # Solo empresa
print(arr_empresas[:, 2])
#Hace que me traiga solo los valores de la empresa
```

```python
arr_empresas[:, 3]      # Solo fecha
# hace que lea todas las filas con ";" y traiga el valor de la columna 3
```

```python
arr_empresas[1:, [2,3]] # Desde segunda fila
```

* ¿Qué representa la expresión arr_empresas[0] dentro del arreglo?
* ¿Qué elemento del arreglo se está seleccionando con arr_empresas[0, 3]?
* ¿Qué filas y columnas selecciona arr_empresas[:, [2,3]]?

```python
#Representa el elemento de la primera fila
print(arr_empresas[0])
```

```python
#me quedo con la fila 0 y la fecha y me traigo solo eso
#basicamente estoy diciendo que del primer elemento (fila 0) me traiga solo la fecha

print(arr_empresas[0, 3])
```

```python
#Selecciona todas las filas para traerme los elemnetos 2 y 3, es decir empresa y fecha
print(arr_empresas[:, [2,3]])
```
