## Resultados interactivos y enriquecidos
Hasta ahora, todos los resultados generados han sido texto, pero pueden ser más interesantes, como el gráfico que aparece a continuación.

`````
import numpy as np
from matplotlib import pyplot as plt

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)

plt.title("Rellenos y ejemplo de alfa")
plt.show()
`````

# 🎯 Integración con Drive

Colaboratory está integrado con Google Drive. Permite compartir, comentar y colaborar en el mismo documento con varias personas:

# 🎯 ¿Qué es Markdown?

Colab tiene dos tipos de celdas: de texto y de código. Las celdas de texto se formatean con un lenguaje de marcado sencillo llamado Markdown.

Para ver el código fuente Markdown, haz doble clic en una celda de texto. Así se mostrarán tanto el código fuente como la versión renderizada. Encima del código fuente Markdown hay una barra de herramientas para facilitar la edición.

Markdown | Preview
--- | ---
`**bold text**` | **bold text**
`*italicized text*` or `_italicized text_` | *italicized text*
`` `Monospace` `` | `Monospace`
`~~strikethrough~~` | ~~strikethrough~~
`[A link](https://www.google.com)` | [A link](https://www.google.com)
`![An image](https://www.google.com/images/rss.png)` | ![An image](https://www.google.com/images/rss.png)

---
